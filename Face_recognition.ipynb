```python
from flask import Flask, request
from werkzeug.utils import secure_filename
import face_recognition

app = Flask(__name__)

@app.route('/upload', methods=['POST'])
def upload_file():
    if 'captured_image' not in request.files or 'saved_image' not in request.files:
        return 'No image files uploaded', 400

    captured_file = request.files['captured_image']
    saved_file = request.files['saved_image']

    if captured_file.filename == '' or saved_file.filename == '':
        return 'Empty file name', 400

    captured_filename = secure_filename(captured_file.filename)
    saved_filename = secure_filename(saved_file.filename)

    # Save uploaded images with unique file names
    captured_path = f'{captured_filename}'
    saved_path = f'{saved_filename}'
    captured_file.save(captured_path)
    saved_file.save(saved_path)

    # Load the images for face recognition
    captured_image = face_recognition.load_image_file(captured_path)
    saved_image = face_recognition.load_image_file(saved_path)

    # Perform face recognition
    captured_faces = face_recognition.face_encodings(captured_image)
    saved_faces = face_recognition.face_encodings(saved_image)

    if captured_faces and saved_faces:
        match = face_recognition.compare_faces(captured_faces, saved_faces[0])
        if match[0]:
            return 'There\'s a match'
        else:
            return 'No match detected'
    else:
        return 'No face detected in one or both of the images'

if __name__ == '__main__':
    app.run(host='0.0.0.0', port=5000)
```
